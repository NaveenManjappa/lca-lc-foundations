```python
from langchain.chat_models import init_chat_model
```

# Intializing and Invoking a model

```python
model = init_chat_model(model="gemini-3.1-flash-lite",model_provider="google_genai")
```

```python
response = model.invoke("Why do parrots talk?")
#print(type(response))
print(response.content)
#print(response.usage_metadata)
```

```python
from pprint import pprint
pprint(response)
```

## Customizing the model

```python
model = init_chat_model(
  model="gemini-3.1-flash-lite",
  model_provider="google_genai",
  temperature=0.5
)

response = model.invoke("What is the capital of moon?")
print(response.content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
response = model.invoke("What is the capital of moon?")
print(response.content)
```

## Initializing and invoking an agent

```python
from langchain.agents import create_agent
agent = create_agent(
    model=model)
```

```python
from langchain.messages import HumanMessage
response = agent.invoke(
  {"messages":[HumanMessage(content="what is the capital of Moon?")]}
)
```

```python
from pprint import pprint
print(response)
```

```python
print(response['messages'][-1].content[0]["text"])
```

```python
from langchain.messages import AIMessage
response = agent.invoke(
  {"messages":[HumanMessage(content="what is the capital of Moon?"), AIMessage(content="The capital of Moon is Moon City."),HumanMessage(content="Interesting,tell me more about Moon City.")]}
)
pprint(response)
```

```python
for token,metadata in agent.stream(
  {"messages":[HumanMessage(content="Tell me all about Moon City, the capital of Moon.")]},
  stream_mode="messages"
):
  if(token.content):
    print(token.content,end="",flush=True)
```
